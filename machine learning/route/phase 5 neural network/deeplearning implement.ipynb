```python
#day19
#TensorFlow implementation:
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
model = Sequential([
    Dense(units=25,activation='sigmoid'),
    Dense(units=15,activation='sigmoid'),
    Dense(units=1,activation='sigmoid')
])#1.how to compute for the inference/prediction
from tensorflow.keras.losses import BinaryCrossentropy
model.compile(loss=BinaryCrossentropy())#2.compile the model using a specific loss function
#loss=MeanSquaredError()可以直接调用不同的损失函数
x=[[200,17]]
y=[0]
model.fit(x,y,epochs=100)#epochs梯度下降步数
#3.train the model
#logistic regression 代码图解见图
```

```python
#training details:
#1.dense函数矩阵+sigmoid归一
#2.交叉熵损失函数：L(f_x->,y)=-ylog(f_x->)-(1-y)log(1-f_x->)
#3.backpropagation in order to compute these partial derivative terms(反向传播算法计算偏导数项)
#多层感知机
```

```python
#day20
#alternatives to the sigmoid activation（sigmoid激活函数替代方案）
#ReLU g(z)=0 if z<0 g(z)=z if z>=0  ->max(0,z)
#线性激活函数：wx+b不变
#softmax激活函数
#choosing activation functions：
#according to output layer：
#Binary classification->sigmoid
#regression->linear activation function
#regression(non-negative)->ReLU
#ReLU common choice（hidden layer）:
#compute faster,one-side flat(j下降更快):gradient decent faster

#why do we need activation function:
#don't use linear regression for hidden layer(一个顶多个线性组合)
#其他激活函数使成为非线性可以更好拟合

#multiclass classification：
#refers to classification problem where you can have more than just 2 possible out put labels(not just 0 or 1)
#多类分类问题：y提前有标签：y1y2y3...
```

```python
#day21
#softmax:
#the softmax regression algorithm is a generalization of logistic regression to the multi-cost classification context
#相比二元都是归一化：公式变为ai=e^zi/(e^zi+e^zj+e^zu+e^zv+...n)
#cost function:见图

#neural network with softmax output:
#put it at out layer
#将输出层神经元每个对应一个分类
#z31=w31*a->+b31
#z32=w32*a->+b32
#z33=w33*a->+b33
#....
#a31=e^z31/(e^z31+e^z32+e^z33+...),a32=...
#这些激活值的每一个都依赖所有z的值
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
model = Sequential([
    Dense(units=25,activation='relu'),
    Dense(units=15,activation='relu'),
    Dense(units=10,activation='softmax')#put it at out layer
])
from tensorflow.keras.losses import SparseCategoricalCrossentropy
model.compile(loss=SparseCategoricalCrossentropy())
#系数分类稀疏交叉熵函数：将y分类为类别，稀疏指y只能取这m个的一个
model.fit(x,y,epochs=100)
#but not recommend to use this version
```

```python
#day22
#improved implementation of softmax:
#不同的计算方式会产生舍入误差
model = Sequential([
    Dense(units=25,activation='relu'),
    Dense(units=15,activation='relu'),
    Dense(units=10,activation='linear')#put it at out layer
])
model.compile(loss=SparseCategoricalCrossentropy(from_logits=True))#from_logits=True表示输出层不使用softmax
#系统高精计算
model.fit(x,y,epochs=100)
logit=model(x)
proba=tf.nn.softmax(logit)

#classification with multiple outputs:
#train one neural network with three outputs
#多分类问题softmax找概率最大的，多标签问题sigmoid（例：每个神经元对应一个label）找概率大于阈值的汇总

#advanced optimization(高级优化)：
#Adam algorithm:
#一种调节学习率的算法
#if wj(or b) keeps moving in same direction,increase alpha
#if wj(or b) keeps oscillating,decrease alpha
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),loss=SparseCategoricalCrossentropy(from_logits=True))
```

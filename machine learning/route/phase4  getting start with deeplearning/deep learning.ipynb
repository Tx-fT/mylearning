```python
#day18
#how neural networks are implemented efficiently:
#vectorization:
x=np.array([[200,17]])
w=np.array([[1,-3,5],
            [-2,4,-6]
])
b=np.array([1,2,3])
def dense(ain,w,b):
    z=np.matmul(ain,w)+b#矩阵乘法
    a=g(z)
    return a
a=dense(x,w,b)
at=a.T#转置
z=at @ w#@就是矩阵乘法
print(z)
```
